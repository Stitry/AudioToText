```python
import pandas as pd
from scipy.stats import chi2_contingency

pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)
pd.set_option("display.max_colwidth", None)
# =========================
# LOAD
# =========================
df = pd.read_csv("BASEfinaleOK(BDD).csv", sep=";", encoding="utf-8-sig")

target = "ageaine"

rows = []

for col in df.columns:
    if col == target:
        continue
    
    sub = df[[target, col]].dropna()
    
    if sub[target].nunique() < 2 or sub[col].nunique() < 2:
        continue
    
    ct = pd.crosstab(sub[target], sub[col])
    
    if ct.shape[0] < 2 or ct.shape[1] < 2:
        continue
    
    chi2, p, dof, _ = chi2_contingency(ct, correction=False)
    
    rows.append({
        "variable": col,
        "n": len(sub),
        "chi2": chi2,
        "ddl": dof,
        "p_value": p
    })

res = pd.DataFrame(rows).sort_values("p_value")

print(res.head(120))
# Nombre de tests significatifs (p < 0.05)
nb_signif = (res["p_value"] < 0.05).sum()

print("Nombre de correspondances avec p < 0.05 :", nb_signif)
res.to_csv("classville_chi2_only.csv", index=False, encoding="utf-8-sig")
```
